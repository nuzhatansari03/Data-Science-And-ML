### Student Report Card 

###### This is a Python mini project that generates report cards for students.

##### Features

###### Stores student data using dictionaries and lists 
###### Calculates total and average marks
###### Assigns grades automatically Checks student attendance
###### Displays a formatted report card

```python
student_list=[
    {'name':'Khushi','roll_no':46,'marks':[85,88,72]},
    {'name':'Pooja','roll_no':48,'marks':[59,69,70]},
    {'name':'Jyoti','roll_no':56,'marks':[84,70,83]},
    {'name':'Kavita','roll_no':60,'marks':[89,92,80]}
]
subject=('Python','Java','Maths')
def report_card():
    for student in student_list:
        present = input(f"Is {student['name']} present? (yes/no): ")
        if present == "no":
            continue
        total_marks=sum(student['marks'])    
        avg_marks=total_marks/len(student['marks'])
        if avg_marks>=90:
            grade='A+'
        elif avg_marks>=85:
            grade='A'
        elif avg_marks>=80:
            grade='B+'    
        elif avg_marks>=75:
            grade='B'
        elif avg_marks>=70:
            grade='C'
        elif avg_marks>=65:
            grade='D'
        else:
            grade='F'    
        print("=" * 50)
        print(f"STUDENT REPORT CARD OF {student['name']}")
        print("=" * 50)
        print(f"Name             : {student['name']}")
        print(f"Roll Number      : {student['roll_no']}")
        for sub, mark in zip(subject, student['marks']):
            print(f"{sub}\t\t : {mark}")  
        print(f"Total Marks      : {total_marks}")
        print(f"Average Marks    : {avg_marks:.2f}")
        print(f"Grade            : {grade}")
        print("=" * 50)
```

```python
report_card()
```
