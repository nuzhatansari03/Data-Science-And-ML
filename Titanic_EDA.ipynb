# **Phase 1: Setup and Inspection**

## **Objective**

In this phase, we will:
- Import the required libraries.
- Load the Titanic dataset.
- Inspect the structure of the dataset.
- Identify missing values before cleaning.

```python
import pandas as pd
import numpy as np

titanic = pd.read_csv("titanic.csv") 

print("========== First 5 Rows ==========")
print(titanic.head())

print("\n========== Dataset Information ==========")
titanic.info()

print("\n========== Missing Values ==========")
print(titanic.isnull())

print("\n========== Number of Missing Values ==========")
print(titanic.isnull().sum()) 
```

# Phase 2: Data Cleaning

## Objective
The dataset contains missing values.
In this phase, we will:
- Remove unnecessary columns.
- Fill missing values.
- Prepare the dataset for analysis.

```python
print("========== Drop Cabin Column ==========")

titanic.drop("Cabin", axis=1, inplace=True) 
print('\n')
```

```python
print("========== Fill Missing Age ==========")

median_age = titanic["Age"].median()

titanic.fillna({"Age": median_age}, inplace=True)
```

```python
print("========== Fill Missing Embarked ==========")

titanic.fillna({"Embarked": "Unknown"}, inplace=True)
```

```python
print("========== Missing Values After Cleaning ==========")

print(titanic.isnull().sum())
```

# Phase 3: Filtering and Hypothesis Testing

## Stakeholder Questions

The stakeholders want to know:

- Did gender affect survival?
- Did age affect survival?
- Did passenger class affect survival?
- How many passengers survived?

```python
children = titanic[titanic["Age"] < 18]

print("Children")
print(children)

print("Total Children:", children.shape[0])
```

```python
survival_rate = titanic["Survived"].mean() * 100

print("Overall Survival Rate")

print(f"{survival_rate:.2f}%")
```

```python
survived = titanic[titanic["Survived"] == 1]

print("Total Survived Passengers")

print(survived.shape[0])
```

```python
survived_children = titanic[
    (titanic["Age"] < 18) &
    (titanic["Survived"] == 1)
]

print("Survived Children")

print(survived_children.shape[0])
```

```python
survived_adults = titanic[
    (titanic["Age"] >= 18) &
    (titanic["Survived"] == 1)
]

print("Survived Adults")

print(survived_adults.shape[0])
```

```python
survived_class1 = titanic[
    (titanic["Pclass"] == 1) &
    (titanic["Survived"] == 1)
]

survived_class2 = titanic[
    (titanic["Pclass"] == 2) &
    (titanic["Survived"] == 1)
]

survived_class3 = titanic[
    (titanic["Pclass"] == 3) &
    (titanic["Survived"] == 1)
]

print("Class 1 Survivors:", survived_class1.shape[0])
print("Class 2 Survivors:", survived_class2.shape[0])
print("Class 3 Survivors:", survived_class3.shape[0])
```

```python
survived_male = titanic[
    (titanic["Sex"] == "male") &
    (titanic["Survived"] == 1)
]

survived_female = titanic[
    (titanic["Sex"] == "female") &
    (titanic["Survived"] == 1)
]

print("Male Survivors:", survived_male.shape[0])

print("Female Survivors:", survived_female.shape[0]) 
```

# Phase 4: GroupBy Analytics

## Objective

In this phase, we use the Split → Apply → Combine methodology to answer stakeholder questions.

We group passengers by:

- Gender
- Passenger Class
- Age Group

Then calculate:
- Total Passengers
- Survivors
- Survival Rate

```python
gender = titanic.groupby("Sex").agg(
    Total_Passengers=("PassengerId", "count"),
    Survivors=("Survived", "sum"),
    Survival_Rate=("Survived", "mean")
)

gender["Survival_Rate"] = gender["Survival_Rate"] * 100

print(gender) 
```

```python
pclass = titanic.groupby("Pclass").agg(
    Total_Passengers=("PassengerId", "count"),
    Survivors=("Survived", "sum"),
    Survival_Rate=("Survived", "mean")
)

pclass["Survival_Rate"] = pclass["Survival_Rate"] * 100

print(pclass)
```

```python
titanic["Age Group"] = np.where(
    titanic["Age"] < 18,
    "Child",
    "Adult"
)

age_group = titanic.groupby("Age Group").agg(
    Total_Passengers=("PassengerId", "count"),
    Survivors=("Survived", "sum"),
    Survival_Rate=("Survived", "mean")
)

age_group["Survival_Rate"] = age_group["Survival_Rate"] * 100

print(age_group)
```
